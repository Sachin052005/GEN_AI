```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")
text = """Artificial Intelligence is transforming the world.
Machine learning and deep learning are popular AI technologies.
Artificial Intelligence helps automate tasks and solve real-world problems."""
doc = nlp(text)
```

```python
words = [token.text.lower()
    for token in doc
    if not token.is_stop and not token.is_punct]

word_frequency = Counter(words)
print("Top 5 Most Frequent Words")
print('-'*25)

for word, count in word_frequency.most_common(5):
    print(f"{word:<15} : {count}")
```

```python
import warnings
warnings.filterwarnings("ignore")
print("Word Similarity")
print('-'*20)
pairs = [
    ("car", "bus"),
    ("doctor", "nurse"),
    ("python", "java"),
    ("apple", "banana"),]
for w1, w2 in pairs:
    doc1 = nlp(w1)
    doc2 = nlp(w2)
    print(f"{w1:10} <-> {w2:10} : {doc1.similarity(doc2):.4f}")
```

```python
print("Sentence Similarity")
print('-'*20)

sentence1 = nlp("Artificial Intelligence is changing the future.")
sentence2 = nlp("AI is transforming the future of technology.")
sentence3 = nlp("I love playing football on weekends.")
print(f"Sentence 1 & Sentence 2 : {sentence1.similarity(sentence2):.4f}")
print(f"Sentence 1 & Sentence 3 : {sentence1.similarity(sentence3):.4f}")
```
