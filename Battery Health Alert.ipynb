```python
import numpy as np
import pandas as pd

Battery=np.array([[40,50,60,70,80,90,95,100]]).reshape(-1,1)
Replace=np.array([1,1,1,0,0,0,0,0])
df=pd.DataFrame({'Battery':Battery.flatten(),'Replace':Replace})
print('Dataset:')
print(df)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(Battery,Replace,test_size=0.3,random_state=42)
print('Training Samples:',len(X_train))
print('Testing Samples:',len(X_test))
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(Battery,Replace)

New_Battery=np.array([[83]])
prediction=model.predict(New_Battery)
probability=model.predict_proba(New_Battery)
print('Battery:',New_Battery[0][0])
print('probability:',probability[0])
```

```python
if prediction[0] == 1:
    print("Prediction: Replace Battery")
else:
    print("Prediction: Battery is Good")

print("Probability:", probability[0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(Battery, Replace, color="blue")
x = np.linspace(40, 100, 100).reshape(-1, 1)
y = model.predict_proba(x)[:, 1]
plt.plot(x, y, color="red")
plt.xlabel("Battery Level")
plt.ylabel("Replace")
plt.title("Battery Replacement Prediction")
plt.grid(True)
plt.show()
```
