```python
import spacy
from spacy.matcher import Matcher

nlp_sm = spacy.load("en_core_web_sm")
!python -m spacy download en_core_web_md
nlp_md = spacy.load("en_core_web_md")
text1 = """Sachin joined TCS as a Software Engineer in Bengaluru.
He worked on a Generative AI project with OpenAI technologies.
Last month, he traveled to Hyderabad to attend a Python conference."""
doc1 = nlp_sm(text1)
```

```python
print("RULE-BASED MATCHING")
print("="*60)
text2 = """Many people start their morning with green tea and mint water.
Some also drink lemon water before breakfast."""
doc2 = nlp_sm(text2)
matcher = Matcher(nlp_sm.vocab)

pattern1 = [{"LOWER": "green"}, {"LOWER": "tea"}]
pattern2 = [{"LOWER": "mint"}, {"LOWER": "water"}]
pattern3 = [{"LOWER": "lemon"}, {"LOWER": "water"}]
matcher.add("Healthy_Drinks", [pattern1, pattern2, pattern3])
matches = matcher(doc2)
for match_id, start, end in matches:
    span = doc2[start:end]
    print("Matched:", span.text)
```

```python
print('WORD VECTORS')
print('-'*30)
text3 = """Samsung has expanded its market overseas,
while Nokia lost its position in the electronics market."""
doc3 = nlp_md(text3)
for token in doc3:
    if not token.is_punct and not token.is_space:
        print(f"\nWord : {token.text}")
        print("Vector Length :", len(token.vector))
        print("First 10 Values :", token.vector[:10])
```

```python
print('WORD SIMILARITY')
print('-'*30)
words = [("Samsung","Nokia"),("Tea","Coffee"),("India","Bengaluru"),("Phone","Mobile"),("Market","Business")]
for w1, w2 in words:
    similarity = nlp_md(w1).similarity(nlp_md(w2))
    print(f"{w1:10} <-> {w2:10} = {similarity:.3f}")
```
