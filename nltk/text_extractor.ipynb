```python
from bs4 import BeautifulSoup
import urllib.request 
import nltk
data = urllib.request.urlopen('https://www.linkedin.com')
content = data.read()
soup = BeautifulSoup(content)
content_data = soup.get_text(strip=True)
s_data = [i for i in content_data.split()]
fq = nltk.FreqDist(s_data) 
for k,v in fq.items():
    print(str(k),':',str(v))
    
```

```python
fq.plot(15)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
data = urllib.request.urlopen('https://linkedin.com')
content = data.read()
soup = BeautifulSoup(content)
text_data =  soup.get_text(strip = True)
tkn=[i for i in text_data.split()]

clear_tkn = tkn[:]
word = stopwords.words('english')
for t in tkn:
    if t in stopwords.words('english'):
        clear_tkn.remove(t)
        
fd = nltk.FreqDist(clear_tkn)
for i,j in fd.items():
    print(str(i),'=',str(j))
```

```python
fd.plot(15)
```
