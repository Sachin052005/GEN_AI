```python
def preprocess(text: str): # here, DRIVER code
    """Lowercase, strip punctuation, tokenize, and lemmatize a string."""
    text = text.lower().translate(str.maketrans("", "", string.punctuation))
    tokens = nltk.word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(tok) for tok in tokens]
    return lemmas
```

```python
# Precompute lemmatized patterns for every intent
processed_patterns = {
    tag: [preprocess(p) for p in data["patterns"]]
    for tag, data in intents.items()
}


def score_intent(user_tokens, pattern_tokens):
    """Overlap score between user tokens and a single pattern's tokens."""
    if not pattern_tokens:
        return 0.0
    overlap = len(set(user_tokens) & set(pattern_tokens))
    return overlap / len(set(pattern_tokens))


def match_intent(user_text, threshold=0.4):
    user_tokens = preprocess(user_text)
    best_tag, best_score = "fallback", 0.0

    for tag, pattern_list in processed_patterns.items():
        if tag == "fallback":
            continue
        for pattern_tokens in pattern_list:
            score = score_intent(user_tokens, pattern_tokens)
            if score > best_score:
                best_score, best_tag = score, tag

    if best_score < threshold:
        return "fallback"
    return best_tag


def get_response(user_text):
    tag = match_intent(user_text)
    return random.choice(intents[tag]["responses"])
```

```python
def speak(text: str, rate: int = 175, volume: float = 1.0):
    """Speak the given text out loud using pyttsx3."""
    engine = pyttsx3.init()
    engine.setProperty("rate", rate)
    engine.setProperty("volume", volume)
    engine.say(text)
    engine.runAndWait()
    engine.stop()
```

```python
def listen(timeout=5, phrase_time_limit=8): # voice chatbot - driver code
    """Record from the microphone and return the transcribed text, or None on failure."""
    recognizer = sr.Recognizer() # recognizes voice and frequency
    with sr.Microphone() as source:
        print("🎤 Listening... (speak now)")
        recognizer.adjust_for_ambient_noise(source, duration=0.5)
        try:
            audio = recognizer.listen(source, timeout=timeout, phrase_time_limit=phrase_time_limit)
        except sr.WaitTimeoutError:
            print("⏱️ Listening timed out — no speech detected.")
            return None

    try:
        text = recognizer.recognize_google(audio)
        print(f"🗣️ You said: {text}")
        return text
    except sr.UnknownValueError:
        print("😕 Could not understand the audio.")
        return None
    except sr.RequestError as e:
        print(f"⚠️ Speech recognition service error: {e}")
        return None
```

```python
def run_voice_chatbot(max_turns=40):
    print("Voice chatbot started. Say something! (say 'bye' to stop)\n")
    speak("Hello! I am your voice assistant. How can I help you?")

    for _ in range(max_turns):
        user_text = listen()
        if user_text is None:
            continue

        tag = match_intent(user_text)
        response = get_response(user_text)
        print(f"🤖 Bot: {response}")
        speak(response)

        if tag == "goodbye":
            break
    else:
        print("Reached max turns for this session.")

run_voice_chatbot()
```
