```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import wordnet

data = """Thanks for this wonderful opportunity. I am Sachin, from Chennai.
I am a Python Developer. I love learning Generative AI."""

print("Sentence Tokenization:")
print(sent_tokenize(data))
```

```python
print("\nWord Tokenization:")
print(word_tokenize(data))
```

```python
french_text = "Bonjour. Bonne après-midi."
print("\nFrench Sentence Tokenization:")
print(sent_tokenize(french_text, language='french'))
```

```python
word = "pleasure"
synsets = wordnet.synsets(word)

print("\nSynsets:")
print(synsets)
print("\nExamples:")
print(synsets[0].examples())
print("\nDefinition:")
print(synsets[0].definition())
```
