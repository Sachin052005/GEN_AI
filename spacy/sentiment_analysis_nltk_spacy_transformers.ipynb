```python
import nltk
import spacy
from spacy.tokens import Doc
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from transformers import pipeline

nltk.download("vader_lexicon")
nlp = spacy.load("en_core_web_sm")
analyzer = SentimentIntensityAnalyzer()
```

```python
def polarity_scores(doc):
    return analyzer.polarity_scores(doc.text)
Doc.set_extension(
    "polarity_scores",
    getter=polarity_scores,
    force=True)

sentences = ["I cannot take this anymore.","Why so serious?",
    "I love learning Generative AI.","This movie is amazing!",
    "I hate waiting in traffic.","The service was okay."]

print("NLTK VADER Sentiment Analysis")
print('-'*20)

for text in sentences:
    doc = nlp(text)
    print(f"\nText: {text}")
    print(doc._.polarity_scores)
```

```python
classifier = pipeline("sentiment-analysis")
print("Transformer Sentiment Analysis")
print('-'* 20)

texts = [
    "I hate cinema.","I love singing.",
    "This phone is fantastic.","The food was terrible.",
    "Today is a wonderful day."]

for text in texts:
    result = classifier(text)
    print(f"\nText: {text}")
    print(result)
```
