```python
import numpy as np
stock = np.array([120, 85, 60, 150, 95, 40, 200, 75])

print("Current Stock:")
print(stock)

print("\nTotal Products in Stock:")
print(np.sum(stock))
print("\nAverage Stock:")
print(np.mean(stock))
print("\nHighest Stock:")
print(np.max(stock))
print("\nLowest Stock:")
print(np.min(stock))
print("\nProducts Below 80 Units:")
print(stock[stock < 80])
updated_stock = np.where(stock < 80, stock + 50, stock)

print("\nUpdated Stock After Restocking:")
print(updated_stock)
print("\nTotal Stock After Restocking:")
print(np.sum(updated_stock))
```
