```python
import warnings
warnings.filterwarnings("ignore")
import spacy
nlp = spacy.load('en_core_web_sm')

word1 = nlp('dog')
word2 = nlp('cat')

print('similarities between', word1.text, 'and', word2.text, 'is', word1.similarity(word2))
```

```python
word1 = nlp('car')
word2 = nlp('bus')

print('similarities between', word1.text, 'and', word2.text, 'is', word1.similarity(word2))
```

```python
word1 = nlp('king')
word2 = nlp('queen')

print('similarities between', word1.text, 'and', word2.text, 'is', word1.similarity(word2))
```

```python
word = nlp('running')

print('Word:', word.text)
print('Lemma:', word[0].lemma_)
```

```python
word = nlp('Python')

print('Word:', word.text)
print('Part of Speech:', word[0].pos_)
```
