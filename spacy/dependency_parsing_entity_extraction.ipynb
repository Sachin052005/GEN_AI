```python
import spacy
nlp = spacy.load("en_core_web_sm")

def flat_subtree(subtree):
    return " ".join([token.text for token in subtree])

text = """Sachin joined TCS as a Software Engineer in Bengaluru.
He worked on a Generative AI project with OpenAI technologies.
Last month, he traveled to Hyderabad to attend a Python conference."""
doc = nlp(text)
```

```python
prepositions = []
for token in doc:
    if token.dep_ == "prep":
        prepositions.append(flat_subtree(token.subtree))
print('PREPOSITIONAL PHRASES')
print('-'*20)
for prep in prepositions:
    print(prep)
```

```python
print("DEPENDENCY PARSING")
print("-"*20)
for token in doc:
    print(f"{token.text:<15} POS={token.pos_:<8} DEP={token.dep_:<10} HEAD={token.head.text}")
```
