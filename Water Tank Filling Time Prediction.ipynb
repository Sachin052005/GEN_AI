```python
import numpy as np

liters = np.array([[100], [200], [300], [400], [500], [600]])
time = np.array([8, 16, 24, 32, 40, 48])
print("Water (Liters):")
print(liters)
print("\nTime (Minutes):")
print(time)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(liters, time)

new_liters = np.array([[750]])
predicted_time = model.predict(new_liters)
print("\nPredicted Filling Time:", round(predicted_time[0], 2), "Minutes")
```

```python
import matplotlib.pyplot as plt
plt.scatter(liters, time, color="blue", label="Training Data")
plt.plot(liters, model.predict(liters), color="red", label="Regression Line")
plt.scatter(new_liters, predicted_time, color="green", s=100, label="Prediction")

plt.title("Water Tank Filling Time Prediction")
plt.xlabel("Water Quantity (Liters)")
plt.ylabel("Filling Time (Minutes)")
plt.legend()
plt.grid(True)
plt.show()
```
