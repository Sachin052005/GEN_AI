```python
!pip install spacy
!python -m spacy download en_core_web_sm
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")
text = """Artificial Intelligence is transforming the world rapidly.
Students are learning AI to build innovative applications.
Python makes Natural Language Processing easier with libraries like spaCy.
Learning every day helps developers improve their skills."""
```

```python
doc = nlp(text)
print("Total Characters :", len(text))
print("Total Tokens     :", len(doc))
print("Total Sentences  :", len(list(doc.sents)))
```

```python
print('-------------------sentence-----------------------------------')
for i in doc.sents:
    print(i)
```

```python
print("-----------TOKENS----------------- ")
for token in doc:
    print(token.text)
```

```python
print('-----------lemmatization----------------')
for i in doc:
    print(i.text,'-->',i.lemma_)
```

```python
print("---------- PARTS OF SPEECH------------------")
for i in doc:
    print(i.text,'--->',i.pos_)
```

```python
print("-------------detailed tags---------------------")
for i in doc:
    print(i.text,'---->',i.tag_)
```

```python
print("---------------STOP WORDS-------------------")
for i in doc:
    if i.is_stop:
        print(i.text)
```

```python
print('----------------------POS SUMMARY--------------------')
pos_list = []
for i in doc:
    pos_list.append(i.pos_)

for i in set(pos_list):
    print(i, "---->", pos_list.count(i))
```

```python
print("-----------------WORDS ONLY------------------")
for i in doc:
    if i.is_alpha:
        print(i.text)
```
