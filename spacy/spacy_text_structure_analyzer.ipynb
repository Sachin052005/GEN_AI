```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")
text = """Artificial Intelligence is transforming the world rapidly.
Students are learning AI to build innovative applications.
Python makes Natural Language Processing easier with libraries like spaCy.
Learning every day helps developers improve their skills."""
```

```python
doc = nlp(text)
nouns=[]
adjectives=[]
for i in doc:
    if i.pos_=='NOUN':
        nouns.append(i.text)
print(nouns)
```

```python
for i in doc:
    if i.pos_=='ADJ':
        adjectives.append(i.text)
print(adjectives)
```

```python
import random
print(random.choice(adjectives)+' '+random.choice(nouns))
```

```python
verbs=[]
for i in doc:
    if i.pos_=='VERB':
        verbs.append(i.text)
print("Verbs:", verbs)
```

```python
pastparticple=[]
for i in doc:
    if i.tag_=='VBN':
        pastparticple.append(i.text)
print("Past Participles (VBN):", pastparticple)
```

```python
for i in doc.noun_chunks:
    print(i.text)    
```

```python
for i in list(doc.sents)[1]:
    print('word:',i.text)
    print('tag:',i.tag_)
    print('head:',i.head.text)
    print('dependency:',i.dep_)
    print('child:',list(i.children))
```
