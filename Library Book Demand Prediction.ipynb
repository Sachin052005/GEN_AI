```python
import numpy as np
import pandas as pd
visitors = np.array([[50], [100], [150], [200], [250], [300]])
books = np.array([40, 80, 120, 160, 200, 240])
print("Data Set")
data=pd.DataFrame({'Visitors':visitors.flatten(),'Books':books})
print(data)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(visitors, books)

new_visitors = np.array([[180]])
prediction = model.predict(new_visitors)
print("Predicted Books Borrowed:", prediction[0])
```

```python
import matplotlib.pyplot as plt
plt.scatter(visitors, books, label="Actual Data")
plt.plot(visitors, model.predict(visitors), label="Regression Line")
plt.xlabel("Visitors")
plt.ylabel("Books Borrowed")
plt.title("Library Book Demand Prediction")
plt.legend()
plt.show()
```
